## <b>Total Sales</b>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from superstore.utilities import Cleaner, FeatureEngineering

try:
    df = pd.read_excel("../Superstore 2019.xls", index_col="Row ID")
except FileNotFoundError:
    print("Change file name to Superstore 2019.xls")

cleaner = Cleaner(df)
df = cleaner.clean()
fe = FeatureEngineering(df)
df = fe.add_columns()

round(float(df["Sales"].sum()), 2)
```

## <b>Total Profit</b>

```python
round(float(df["Profit"].sum()), 2)
```

## <b>Average Order Value</b>

```python
aov = df.groupby(["Order ID"])["Sales"].sum().mean()
round(float(aov), 2)
```

## <b>Average Shipping Duration</b>

```python
average_shipping_duration = df["Shipping Duration"].dt.days.mean()
round(float(average_shipping_duration))
```

## <b>Top Category (by Sales)</b>

```python
tc = df.groupby(["Category"])["Sales"].sum()
filter = tc == tc.max()
tc.idxmax() 
```

## <b>Top Region (by Sales)</b>

```python
tr = df.groupby(["Region"])["Sales"].sum()
filter = tr == tr.max()
tr.idxmax() 
```

## <b>Average Discount</b>

```python
ad = df["Discount"].mean()
round(float(ad), 2)
```
