## <b>Duplicates</b>

```python
import pandas as pd
try:
    df = pd.read_excel("../Superstore 2019.xls", index_col="Row ID")
except FileNotFoundError:
    raise FileNotFoundError("Change file name to Superstore 2019.xls")

df[df.duplicated()] # 1 duplicate found
```

```python
df.loc[3406:3408] # Row 3406 & 3407 are the same
```

```python
df.drop_duplicates(inplace=True) # Removes 3407 and updates df
df.loc[3406:3408] # 3407 is now gone
```

## <b>Incorrect dtypes</b>

```python
df.info() # Postal Code's Dtype is float64, should be changed to int64 to avoid .0
```

```python
df["Postal Code"] = df["Postal Code"].astype('Int64') # Changes Dtype
df.info() # Successfully changed
```

## <b>Missing Values</b>

```python
df.info() # Postal Code has 9982 non-null values instead of 9993
```

```python
null_postal_codes = df["Postal Code"].isna() #filter

""" 
Notice how all null postal codes come from:
City: Burlington
State: Vermont

Since all these records share the same city & state; we can replace all null values to their postal codes (05401)
"""

df[null_postal_codes] 
```

```python
df["Postal Code"] = df["Postal Code"].fillna(5401) # Default for null becomes 5401
df.info() # No null values
```

## <b>Outliers</b><br>
Outlier detection and removal using IQR method<br><br>
<b>Rules:</b><br>
IQR = Q3 - Q1 (75%, 25% shown in .describe())<br>
Lower Limit: (Q3 * 1.5) - IQR<br>
Upper Limit: (Q1 * 1.5) - IQR<br><br>
Anything above/lower these limits is an outlier<br><br>
Source: https://www.youtube.com/watch?v=A3gClkblXK8
