## <b>Profit Margin</b>

```python
import pandas as pd
from superstore import utilities

try:
    df = pd.read_excel("../Superstore 2019.xls", index_col="Row ID")
except FileNotFoundError:
    raise FileNotFoundError("Change file name to Superstore 2019.xls")

cleaner = utilities.Cleaner(df)
df = cleaner.clean()
df.head(1) # 0 New colums added
```

```python
df["Profit Margin"] = (df["Profit"] / df["Sales"]).round(2) # Adds Profit Margin column by dividing Profit & Sales
```

## <b>Shipping Duration</b>

```python
df["Shipping Duration"] = df["Ship Date"] - df["Order Date"] # Adds Shipping Duration column by subtracting Ship Date & Order Date
```

## <b>Sales Performance Category</b>

```python
"""
I thought about using the sales average (mean) to create the bins, but due to skew data using quartiles is a much better option. 

Sources: 
https://www.youtube.com/watch?v=MJt6NotH4fY
https://pandas.pydata.org/docs/reference/api/pandas.cut.html
"""

cutoff = [
            df["Sales"].quantile(0), 
            df["Sales"].quantile(0.25), 
            df["Sales"].quantile(0.50), 
            df["Sales"].quantile(0.75), 
            df["Sales"].quantile(1)
        ]

labels = ["Low", "Medium", "High", "Very High"] # 4 categories

df["Sales Performance Category"] = pd.cut(df["Sales"], bins=cutoff, labels=labels, include_lowest=True) 
```
